## Nested Data Structures

```python
from faker import Faker

faker = Faker()
```

```python
employees = [
    faker.simple_profile() for i in range(10)
]

employees
```

```python
employees[1]["name"]
```

```python
for profile in employees: 
    if profile["name"] == "Monique Price": print(profile)
```
