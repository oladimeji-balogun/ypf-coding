```python
def area_of_circle(r): 
    area = (22 / 7) * r * r
    return area
```

```python
def perimeter_of_circle(r): 
    perimeter = 2 * (22 / 7) * r
    return perimeter
```

```python
radius_a = 7

area_a = area_of_circle(radius_a)
perimeter_a = perimeter_of_circle(radius_a)

print(f"area of circle a: {area_a}")
print(f"perimeter of circle a: {perimeter_a}")
```
