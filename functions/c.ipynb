```python
def add(*numbers):

    result = 0
    for number in numbers: 
        result = result + number
    return result

    
```

```python
def factorial(n: int) -> int: 
    if n == 0: return 1 

    return n * factorial(n - 1)
```

```python
factorial(4)
```

```python
factorial(5)
```

```python
factorial(6)
```
