## object oriented programming

```python
"""
animal: 
    properties:
        - gender
        - color
        - height
        - name
    
    behaviours: 
        - walk
        - make_sound
        - run 

"""
```

```python
"""
properties: 
    - name
    - age
    - class
    - religion
    - parent's name
    - address
    - dob
    - parent's phone number

behaviours: 
    - read
    - write
    - run
    - jump
    - ask question
    - make noise 

"""

class Student: 
    def __init__(
            self, 
            name: str, 
            age: int, 
            class_: str, 
            religion: str, 
            parents_name: str, 
            address: str, 
            dob: str,
            parents_phone_number: str
    ): 
        self.name = name
        self.age = age
        self.class_ = class_
        self.religion = religion
        self.parents_name = parents_name
        self.address = address 
        self.dob = dob 
        self.parents_phone_number = parents_phone_number

    
    def read(self, book_title: str):
        print(f"{self.name} is reading {book_title}")

    def write(self, where: str): 
        print(f"{self.name} is writing on the {where}")

    def run(self, where: str): 
        print(f"{self.name} is running on the {where}")

    def jump(self): 
        print(f"{self.name} is jumping")

    def run(self): 
        print(f"{self.name} is running.")

    def ask_question(self): 
        print(f"{self.name} is making noise")

    
    
```

```python
ola = Student(
    name="Ola", 
    age=10, 
    class_="jss one", 
    religion="christianity", 
    parents_phone_number="7041348987", 
    parents_name="mr and mrs Ola", 
    address="gbagada", 
    dob="15, march, 2026", 
)
```

```python
ola.name, ola.age, ola.class_, ola.religion, ola.parents_phone_number, ola.parents_name, ola.address, ola.dob
```

```python
ola.read("game of thrones")
```

```python
ola.write(where="chalkboard")
```

```python
ola.run()
```

## inheritance

```python
class Animal: 

    def move(self): 
        pass

    def make_sound(self): 
        pass 

    def eat(self): 
        pass 
```

```python
class Chicken(Animal): 

    def __init__(self, name: str): 
        self.name = name 

    def move(self): 
        print(f"{self.name} is moving")

    def make_sound(self):
        print(f"{self.name} is doing kukurukuuuuuuuu")

    def eat(self): 
        print(f"{self.name} is eating agbado")
```

```python
class Goat(Animal): 
    def __init__(self, name: str): 
        self.name = name

    def move(self): 
        print(f"{self.name} is moving")

    def make_sound(self):
        print(f"{self.name} is bleatinggg")

    def eat(self): 
        print(f"{self.name} is eating leaves")
```

```python
chicken = Chicken(name="lucky")
goat = Goat(name="Tom")
```

```python
chicken.make_sound()
```

```python
goat.make_sound()
```
