```python
class User: 
    def __init__(
            self, 
            name: str, 
            age: int, 
            phone_number: str
    ): 
        self.name = name 
        self.age = age 
        self.phone_number = phone_number

    def get_name(self) -> str: 
        return self.name 

    def get_age(self) -> int: 
        return self.age

    def get_phone_number(self) -> str: 
        return self.phone_number

    def set_name(self, new_name: str): 
        self.name = new_name

    def set_age(self, new_age: int): 
        self.age = new_age

    def set_phone_number(self, new_phone_number: str): 
        self.phone_number = new_phone_number

    def save(self, file_name: str): 
        with open(file_name, "a") as file: 
            file.write(
                f"{self.name} | {self.age} | {self.phone_number}\n"
            )


```

```python
print("=" * 20)
print("you're welcome to our expense tracker application, create an account below")
print("=" * 20)

name = input("enter your name here: ")
age = int(input("enter your age here: "))
phone_number = input("enter your phone number here: ")

user = User(name=name, age=age, phone_number=phone_number)
user.save("../users.txt")
```
